**How to make a cluster using Python**

```python
import pandas as pd
import numpy as np
import scipy.stats as stats
import matplotlib.pyplot as plt
import seaborn as sns
import altair as alt


from scipy.cluster.hierarchy import dendrogram, linkage
from scipy.spatial.distance import pdist
from scipy.cluster.hierarchy import cut_tree
```

Dataset with 6 people (age, income)

```python
data_income= pd.DataFrame(
    {
        'income':[9.60,8.40,2.40,18.20,3.90,6.40],
        'age':[28,31,42,38,25,41]
    }
)

data_income
```

```python
#plot scatter plot
fig, ax = plt.subplots()
fig.set_size_inches(10,6)
ax.scatter(data_income.age, data_income.income, s =0)
for i, txt in enumerate(data_income.index):
    ax.annotate(txt, (data_income.age[i],data_income.income[i]))
plt.hlines(0,0,data_income.age.max(), color = 'gray')
plt.vlines(0,0,data_income.income.max(), color = 'gray')
```
